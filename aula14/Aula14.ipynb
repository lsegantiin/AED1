```python
def gerar_relatorio_financeiro(receita, despesas, imposto=0.15, mostrar_resumo=True):

    lucro_bruto = receita - despesas

    if lucro_bruto <= 0:
        return "Prejuízo registrado. Sem aplicação de imposto"

    imposto_aplicado = lucro_bruto * imposto
    lucro_liquido = lucro_bruto - imposto_aplicado

    if mostrar_resumo:
        print("RELATÓRIO FINANCEIRO DO MÊS")
        print(f"Receita Total: R$ {receita:.2f}")
        print(f"Despesas Totais: R$ {despesas:.2f}")
        print(f"Lucro Bruto: R$ {lucro_bruto:.2f}")
        print(f"Imposto Aplicado ({int(imposto * 100)}%): R$ {imposto_aplicado:.2f}")
        print(f"Lucro Líquido: R$ {lucro_liquido:.2f}")

    return lucro_liquido
```

```python
receita = float(input("Digite o valor da receita: R$ "))
despesas = float(input("Digite o valor das despesas: R$ "))

imposto = float(input("Digite a porcentagem do imposto (ex: 0.15 para 15%): "))

mostrar = input("Deseja mostrar o resumo? (S/N): ").upper()

if mostrar == "S":
    mostrar_resumo = True
else:
    mostrar_resumo = False


# Chamada da função
resultado = gerar_relatorio_financeiro(
    receita,
    despesas,
    imposto,
    mostrar_resumo
)

print("\nResultado:", resultado)
```

```python
def identificar_regiao(estado):
    """
    Função para calcular serviços de entrega de uma loja na internet.
    """
    estado = estado.upper()

    norte = ["AC", "AP", "AM", "PA", "RO", "RR", "TO"]
    nordeste = ["AL", "BA", "CE", "MA", "PB", "PE", "PI", "RN", "SE"]
    centro_oeste = ["DF", "GO", "MT", "MS"]
    sudeste = ["ES", "MG", "RJ", "SP"]
    sul = ["PR", "RS", "SC"]

    if estado in norte:
        return "N"

    elif estado in nordeste:
        return "NE"

    elif estado in centro_oeste:
        return "CO"

    elif estado in sul:
        return "S"

    elif estado in sudeste:
        return "SE"

    else:
        return "Estado inválido"


# Serviços de Logística

def valor_de_entrega(peso, estado, presente=False):

    # Peso do produto
    if peso <= 500:
        valor_peso = 15.00

    elif peso <= 2000:
        valor_peso = 43.00

    else:
        valor_peso = 103.00

    # Identifica a região pelo estado
    regiao = identificar_regiao(estado)

    # Acréscimo por região
    if regiao in ["N", "NE"]:
        valor_regiao = 50

    elif regiao in ["CO", "S"]:
        valor_regiao = 25

    elif regiao == "SE":
        valor_regiao = 0

    else:
        return "Estado inválido"

    # Taxa para presente
    taxa_presente = 10 if presente else 0

    # Valor total
    total = valor_peso + valor_regiao + taxa_presente

    return f"Valor total da entrega: R$ {total:.2f}"
```

```python
peso = float(input("Digite o peso do produto em gramas: "))
estado = input("Digite a sigla do estado de destino: ")

presente_input = input("É presente? (S/N): ").upper()

if presente_input == "S":
    presente = True
else:
    presente = False

# Resultado
print(valor_de_entrega(peso, estado, presente))
```

```python
# Jogo Virtual

import random

# Opções do jogo
opcoes = ["pedra", "papel", "tesoura"]

# Nome do jogador
nome = input("Digite seu nome: ")

vitorias_jogador = 0
vitorias_computador = 0
empates = 0
rodadas = 0

print(f"\nBem-vindo ao Pedra, Papel e Tesoura, {nome}!")

# Loop de 3 jogadas
while rodadas < 3:

    print(f"\nRodada {rodadas + 1}")

    # Escolha do jogador
    jogador = input("Escolha Pedra, Papel ou Tesoura: ").lower()

    # Validação
    if jogador not in opcoes:
        print("Opção inválida!")
        continue

    # Escolha do computador
    computador = random.choice(opcoes)

    print(f"{nome} escolheu: {jogador}")
    print(f"Computador escolheu: {computador}")

    # Regras do jogo
    if jogador == computador:
        print("Empate!")
        empates += 1

    elif (
        (jogador == "pedra" and computador == "tesoura") or
        (jogador == "papel" and computador == "pedra") or
        (jogador == "tesoura" and computador == "papel")
    ):
        print(f"{nome} venceu a rodada!")
        vitorias_jogador += 1

    else:
        print("Computador venceu a rodada!")
        vitorias_computador += 1

    rodadas += 1


# Resultado final
print("\n===== RESULTADO FINAL =====")
print(f"Vitórias de {nome}: {vitorias_jogador}")
print(f"Vitórias do Computador: {vitorias_computador}")
print(f"Empates: {empates}")

# Campeão
if vitorias_jogador > vitorias_computador:
    print(f"{nome} foi o campeão!")

elif vitorias_computador > vitorias_jogador:
    print("O Computador foi o campeão!")

else:
    print("O jogo terminou empatado!")
```
