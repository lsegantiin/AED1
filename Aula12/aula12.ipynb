```python
cidades = []

while True:
    cidade = input("Digite o nome da cidade ou sair: ")
    if cidade.lower() == "sair":
        break
    cidades.append(cidade)

cidades.sort()

print("Relação de cidades em ordem crescente:")
for cidade in cidades:
    print(cidade)
```

```python
from random import randint

frequencias = [0, 0, 0, 0, 0, 0]

for i in range(6000):
    face = randint(1, 6)
    frequencias[face - 1] += 1

print("Frequência das faces após 6000 jogadas:")
for i in range(6):
    p = (frequencias[i] / 6000) * 100
    print(f"Face {i+1}: {frequencias[i]} vezes - {p:6.2f}%")
```
